```python
from tensorflow.keras.preprocessing import image
```

```python
img = image.load_img("img.png", target_size=(224,224))
```

```python
img_array=image.img_to_array(img)
```

```python
img_array.shape
```

```python
img_array.dtype
```

```python
import numpy as np
batched_image = np.expand_dims(img_array, axis=0)
```

```python
print(batched_image.shape)
```

```python
from tensorflow.keras.layers import GlobalMaxPooling2D, Dense, Dropout, Flatten
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input
```

```python
import tensorflow
model = ResNet50(weights='imagenet',include_top=False, input_shape=(224, 224, 3))
model.trainable = False
model=tensorflow.keras.Sequential([
    model,
    GlobalMaxPooling2D(),])
```

```python
preprocessed_img=preprocess_input(batched_image)
```

```python
print(preprocessed_img.shape)
```

```python
model.predict(preprocessed_img)
```

```python
model.predict(preprocessed_img).flatten()
```

```python
from numpy.linalg import norm
model.predict(preprocessed_img).flatten()/norm(model.predict(preprocessed_img).flatten())
```

```python
norm(model.predict(preprocessed_img).flatten())
```
